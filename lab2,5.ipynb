```python
jug1 = 4
jug2 = 3
target = 2

x = 0
y = 0

while x != target and y != target:
    if x == 0:
        x = jug1
    elif y < jug2:
        pour = min(x, jug2 - y)
        x = x - pour
        y = y + pour
    else:
        y = 0

    print("Jug1:", x, "Jug2:", y)

print("Target reached!")
```

```python
5. Write the Python program for Vacuum Cleaner problem.
```

```python

rooms = {'A': 1, 'B': 1}
location = 'A'

for room in rooms:
    location = room
    print("Vacuum is in room", location)

    if rooms[room] == 1:
        print("Room is dirty")
        rooms[room] = 0
        print("Room cleaned")
    else:
        print("Room is clean")

print("Final State:", rooms)
print("All rooms cleaned!")
```
