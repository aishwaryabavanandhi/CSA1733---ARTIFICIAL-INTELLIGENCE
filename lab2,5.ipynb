```python
from collections import deque

def solve_water_jug(jug1_cap, jug2_cap, target):
    # Set to store visited states to avoid infinite loops
    visited = set()

    # Queue for BFS storing tuples of (jug1, jug2, path_taken)
    queue = deque([(0, 0, [])])

    while queue:
        jug1, jug2, path = queue.popleft()

        # If we reached the target in either of the jugs, return the path
        if jug1 == target or jug2 == target:
            return path + [(jug1, jug2)]

        # If this state has already been visited, skip it
        if (jug1, jug2) in visited:
            continue
        visited.add((jug1, jug2))

        # Generate next possible states
        next_states = [
            (jug1_cap, jug2, "Fill Jug 1"),                         # Fill Jug 1
            (jug1, jug2_cap, "Fill Jug 2"),                         # Fill Jug 2
            (0, jug2, "Empty Jug 1"),                               # Empty Jug 1
            (jug1, 0, "Empty Jug 2"),                               # Empty Jug 2
            # Pour Jug 1 -> Jug 2
            (max(0, jug1 - (jug2_cap - jug2)), min(jug2_cap, jug2 + jug1), "Pour Jug 1 -> Jug 2"),
            # Pour Jug 2 -> Jug 1
            (min(jug1_cap, jug1 + jug2), max(0, jug2 - (jug1_cap - jug1)), "Pour Jug 2 -> Jug 1")
        ]

        for next_j1, next_j2, action in next_states:
            if (next_j1, next_j2) not in visited:
                queue.append((next_j1, next_j2, path + [(jug1, jug2, action)]))

    return None

# Example execution:
# Jug 1 capacity = 4, Jug 2 capacity = 3, Target water = 2 liters
jug1_capacity = 4
jug2_capacity = 3
target_water = 2

solution = solve_water_jug(jug1_capacity, jug2_capacity, target_water)

if solution:
    print(f"Steps to measure {target_water} liters using {jug1_capacity}L and {jug2_capacity}L jugs:\n")
    for step in solution[:-1]:
        print(f"Current: (Jug1: {step[0]}L, Jug2: {step[1]}L) -> Action: {step[2]}")
    print(f"Final State reached: (Jug1: {solution[-1][0]}L, Jug2: {solution[-1][1]}L)")
else:
    print("No solution is possible.")
```

```python
5. Write the Python program for Vacuum Cleaner problem.
```

```python
class VacuumCleanerEnvironment:
    def __init__(self):
        # 0: Clean, 1: Dirty
        # Let's define a simple two-room environment (A and B)
        self.locationCondition = {'A': '1', 'B': '1'}
        self.vacuumLocation = 'A'

    def get_status(self):
        return self.locationCondition[self.vacuumLocation]

    def clean(self):
        self.locationCondition[self.vacuumLocation] = '0'

    def move(self, destination):
        self.vacuumLocation = destination


def run_vacuum_cleaner():
    env = VacuumCleanerEnvironment()
    print("Initial State:", env.locationCondition)
    print(f"Vacuum starting at Location: {env.vacuumLocation}\n")

    # Traverse and clean rooms
    for room in ['A', 'B']:
        env.move(room)
        print(f"Vacuum moved to Location: {room}")

        if env.get_status() == '1':
            print(f"Location {room} is Dirty.")
            env.clean()
            print(f"Cleaned Location {room}.")
        else:
            print(f"Location {room} is already Clean.")

    print("\nFinal State:", env.locationCondition)
    print("All rooms have been successfully cleaned!")

run_vacuum_cleaner()
```
